# **1. Vectors**

## What is a Vector?
A **vector** is a mathematical object that has both **magnitude** (length) and **direction**. In data science, vectors are used to represent data points, features, weights, and much more.

Vectors can be thought of in multiple ways:

| Perspective | Description | Example |
|-------------|-------------|--------|
| **Physics** | An arrow with magnitude and direction | Force, velocity |
| **Computer Science** | An ordered list of numbers | `[3, 4, 5]` |
| **Mathematics** | An element of a vector space | $\vec{v} \in \mathbb{R}^n$ |

## Why Vectors Matter in Data Science
- **Feature vectors**: Each data point is a vector of features
- **Weight vectors**: Model parameters in ML are often vectors
- **Embeddings**: Word2Vec, image embeddings are vectors
- **Gradients**: Direction of steepest ascent in optimization

## Vector Notation

A vector $\vec{v}$ in $\mathbb{R}^n$ is written as:

$$\vec{v} = \begin{bmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{bmatrix}$$

For a 2D vector:

$$\vec{v} = \begin{bmatrix} x \\ y \end{bmatrix}$$

```python
# ===============================================
# CREATING VECTORS WITH NUMPY
# ===============================================

import numpy as np
import matplotlib.pyplot as plt

print("=" * 50)
print("CREATING VECTORS")
print("=" * 50)

# A 2D vector
v = np.array([3, 2])
print(f"\n2D Vector v = {v}")
print(f"  Shape: {v.shape}")
print(f"  Dimension: {v.ndim}")

# A 3D vector
w = np.array([1, 4, -2])
print(f"\n3D Vector w = {w}")
print(f"  Shape: {w.shape}")

# Column vector (matrix form)
v_col = np.array([[3], [2]])
print(f"\nColumn vector:\n{v_col}")
print(f"  Shape: {v_col.shape}")
```

```python
# ===============================================
# VISUALIZING VECTORS IN 2D
# ===============================================

print("=" * 50)
print("VISUALIZING VECTORS")
print("=" * 50)

v = np.array([3, 2])
w = np.array([-1, 4])

fig, ax = plt.subplots(figsize=(8, 6))

# Plot vectors as arrows from the origin
ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
          color='steelblue', linewidth=2, label=f'v = {v}')
ax.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1,
          color='coral', linewidth=2, label=f'w = {w}')

# Label the vectors
ax.annotate(f'v = ({v[0]}, {v[1]})', xy=(v[0], v[1]),
            xytext=(v[0]+0.3, v[1]+0.3), fontsize=12, color='steelblue')
ax.annotate(f'w = ({w[0]}, {w[1]})', xy=(w[0], w[1]),
            xytext=(w[0]+0.3, w[1]+0.3), fontsize=12, color='coral')

# Style the plot
ax.set_xlim(-3, 5)
ax.set_ylim(-1, 6)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Vectors in 2D Space')
ax.legend()
plt.tight_layout()
plt.show()
```

## Vector Addition

Vectors are added **component-wise**:

$$\vec{v} + \vec{w} = \begin{bmatrix} v_1 + w_1 \\ v_2 + w_2 \end{bmatrix}$$

Geometrically, this follows the **parallelogram rule** — place the tail of $\vec{w}$ at the tip of $\vec{v}$.

### Properties of Vector Addition

| Property | Formula |
|----------|--------|
| Commutative | $\vec{v} + \vec{w} = \vec{w} + \vec{v}$ |
| Associative | $(\vec{u} + \vec{v}) + \vec{w} = \vec{u} + (\vec{v} + \vec{w})$ |
| Identity | $\vec{v} + \vec{0} = \vec{v}$ |
| Inverse | $\vec{v} + (-\vec{v}) = \vec{0}$ |

```python
# ===============================================
# VECTOR ADDITION
# ===============================================

print("=" * 50)
print("VECTOR ADDITION")
print("=" * 50)

v = np.array([3, 2])
w = np.array([1, 3])
result = v + w

print(f"\nv = {v}")
print(f"w = {w}")
print(f"v + w = {result}")

# Visualize the addition (parallelogram rule)
fig, ax = plt.subplots(figsize=(8, 6))

# Original vectors
ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
          color='steelblue', linewidth=2, label=f'v = {v}')
ax.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1,
          color='coral', linewidth=2, label=f'w = {w}')

# w placed at tip of v (tail-to-tip method)
ax.quiver(v[0], v[1], w[0], w[1], angles='xy', scale_units='xy', scale=1,
          color='coral', linewidth=1.5, linestyle='dashed', alpha=0.5)

# v placed at tip of w
ax.quiver(w[0], w[1], v[0], v[1], angles='xy', scale_units='xy', scale=1,
          color='steelblue', linewidth=1.5, linestyle='dashed', alpha=0.5)

# Resultant vector
ax.quiver(0, 0, result[0], result[1], angles='xy', scale_units='xy', scale=1,
          color='green', linewidth=2.5, label=f'v + w = {result}')

ax.set_xlim(-1, 6)
ax.set_ylim(-1, 7)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_title('Vector Addition (Parallelogram Rule)')
ax.legend()
plt.tight_layout()
plt.show()
```

## Scalar Multiplication (Scaling)

Multiplying a vector by a scalar **scales** its magnitude:

$$c \cdot \vec{v} = \begin{bmatrix} c \cdot v_1 \\ c \cdot v_2 \end{bmatrix}$$

| Scalar | Effect |
|--------|-------|
| $c > 1$ | Stretches the vector |
| $0 < c < 1$ | Shrinks the vector |
| $c = -1$ | Reverses direction |
| $c = 0$ | Produces the zero vector |

```python
# ===============================================
# SCALAR MULTIPLICATION
# ===============================================

print("=" * 50)
print("SCALAR MULTIPLICATION")
print("=" * 50)

v = np.array([2, 1])
scalars = [2, 0.5, -1, 3]

for s in scalars:
    print(f"  {s} * {v} = {s * v}")

# Visualization
fig, ax = plt.subplots(figsize=(8, 6))

colors = ['steelblue', 'coral', 'green', 'purple']
for s, c in zip(scalars, colors):
    scaled = s * v
    ax.quiver(0, 0, scaled[0], scaled[1], angles='xy', scale_units='xy', scale=1,
              color=c, linewidth=2, label=f'{s} × v = {scaled}')

# Original vector (black dashed)
ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
          color='black', linewidth=2, linestyle='dashed', label=f'v = {v}', alpha=0.5)

ax.set_xlim(-4, 8)
ax.set_ylim(-3, 5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_title('Scalar Multiplication of a Vector')
ax.legend()
plt.tight_layout()
plt.show()
```

## Vector Magnitude (Norm) and Direction

The **magnitude** (or **norm**) of a vector is its length. For a 2D vector $\vec{v} = [v_1, v_2]$:

$$\|\vec{v}\| = \sqrt{v_1^2 + v_2^2}$$

For an $n$-dimensional vector:

$$\|\vec{v}\| = \sqrt{\sum_{i=1}^{n} v_i^2}$$

A **unit vector** has magnitude 1. To normalize a vector:

$$\hat{v} = \frac{\vec{v}}{\|\vec{v}\|}$$

```python
# ===============================================
# VECTOR MAGNITUDE AND UNIT VECTORS
# ===============================================

from numpy.linalg import norm

print("=" * 50)
print("VECTOR MAGNITUDE AND UNIT VECTORS")
print("=" * 50)

v = np.array([3, 4])
magnitude = norm(v)
print(f"\nVector v = {v}")
print(f"||v|| = √(3² + 4²) = √(9 + 16) = √25 = {magnitude}")

# Unit vector
v_hat = v / magnitude
print(f"\nUnit vector v̂ = v / ||v|| = {v_hat}")
print(f"||v̂|| = {norm(v_hat):.4f}  (should be 1.0)")

# 3D example
w = np.array([1, 2, 3])
print(f"\n3D Vector w = {w}")
print(f"||w|| = √(1² + 2² + 3²) = √14 = {norm(w):.4f}")
print(f"Unit vector ŵ = {w / norm(w)}")
```

## Dot Product

The **dot product** (or inner product) of two vectors is a scalar:

$$\vec{v} \cdot \vec{w} = \sum_{i=1}^{n} v_i \cdot w_i = v_1 w_1 + v_2 w_2 + \ldots + v_n w_n$$

Geometric interpretation:

$$\vec{v} \cdot \vec{w} = \|\vec{v}\| \cdot \|\vec{w}\| \cdot \cos(\theta)$$

where $\theta$ is the angle between the two vectors.

### Key Insights from the Dot Product

| Condition | Meaning |
|-----------|--------|
| $\vec{v} \cdot \vec{w} > 0$ | Vectors point in similar direction ($\theta < 90°$) |
| $\vec{v} \cdot \vec{w} = 0$ | Vectors are **orthogonal** (perpendicular, $\theta = 90°$) |
| $\vec{v} \cdot \vec{w} < 0$ | Vectors point in opposite-ish directions ($\theta > 90°$) |

```python
# ===============================================
# DOT PRODUCT AND ANGLE BETWEEN VECTORS
# ===============================================

print("=" * 50)
print("DOT PRODUCT")
print("=" * 50)

v = np.array([3, 4])
w = np.array([1, 2])

# Method 1: np.dot
dot_product = np.dot(v, w)
print(f"\nv = {v}, w = {w}")
print(f"v · w = (3)(1) + (4)(2) = 3 + 8 = {dot_product}")

# Method 2: @ operator
print(f"v @ w = {v @ w}")

# Finding the angle between vectors
cos_theta = dot_product / (norm(v) * norm(w))
theta_radians = np.arccos(cos_theta)
theta_degrees = np.degrees(theta_radians)

print(f"\ncos(θ) = {cos_theta:.4f}")
print(f"θ = {theta_degrees:.2f}°")

# Orthogonal vectors example
print("\n" + "-" * 50)
print("ORTHOGONAL VECTORS")
print("-" * 50)

a = np.array([1, 0])
b = np.array([0, 1])
print(f"\na = {a}, b = {b}")
print(f"a · b = {np.dot(a, b)} → Orthogonal! (90°)")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Non-orthogonal
ax = axes[0]
ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
          color='steelblue', linewidth=2, label=f'v = {v}')
ax.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1,
          color='coral', linewidth=2, label=f'w = {w}')
ax.set_xlim(-1, 5)
ax.set_ylim(-1, 5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_title(f'v · w = {dot_product}, θ = {theta_degrees:.1f}°')
ax.legend()

# Plot 2: Orthogonal
ax = axes[1]
ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1,
          color='steelblue', linewidth=2, label=f'a = {a}')
ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1,
          color='coral', linewidth=2, label=f'b = {b}')
ax.set_xlim(-1, 2)
ax.set_ylim(-1, 2)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_title(f'a · b = {np.dot(a, b)} → Orthogonal (90°)')
ax.legend()

plt.tight_layout()
plt.show()
```

## Cross Product (3D only)

The **cross product** of two 3D vectors produces a third vector that is **perpendicular** to both:

$$\vec{v} \times \vec{w} = \begin{bmatrix} v_2 w_3 - v_3 w_2 \\ v_3 w_1 - v_1 w_3 \\ v_1 w_2 - v_2 w_1 \end{bmatrix}$$

The magnitude of the cross product equals the area of the parallelogram formed by the two vectors:

$$\|\vec{v} \times \vec{w}\| = \|\vec{v}\| \cdot \|\vec{w}\| \cdot \sin(\theta)$$

```python
# ===============================================
# CROSS PRODUCT (3D)
# ===============================================

print("=" * 50)
print("CROSS PRODUCT")
print("=" * 50)

v = np.array([1, 0, 0])
w = np.array([0, 1, 0])

cross = np.cross(v, w)
print(f"\nv = {v}")
print(f"w = {w}")
print(f"v × w = {cross}")
print(f"(The cross product is perpendicular to both v and w)")

# Verify perpendicularity
print(f"\nVerification:")
print(f"  cross · v = {np.dot(cross, v)} (should be 0)")
print(f"  cross · w = {np.dot(cross, w)} (should be 0)")

# Another example
a = np.array([2, 3, 4])
b = np.array([5, 6, 7])
c = np.cross(a, b)

print(f"\na = {a}")
print(f"b = {b}")
print(f"a × b = {c}")
print(f"  c · a = {np.dot(c, a)} (should be 0)")
print(f"  c · b = {np.dot(c, b)} (should be 0)")
```

## Linear Combinations and Span

A **linear combination** of vectors $\vec{v}_1, \vec{v}_2, \ldots, \vec{v}_k$ is any vector of the form:

$$c_1 \vec{v}_1 + c_2 \vec{v}_2 + \ldots + c_k \vec{v}_k$$

where $c_1, c_2, \ldots, c_k$ are scalars.

The **span** of a set of vectors is the set of ALL possible linear combinations of those vectors.

### Linear Independence

Vectors are **linearly independent** if no vector in the set can be written as a linear combination of the others.

| Concept | Description |
|---------|------------|
| **Linearly independent** | No redundancy; each vector adds a new "dimension" |
| **Linearly dependent** | At least one vector is redundant |
| **Basis** | A minimal set of linearly independent vectors that spans the space |

```python
# ===============================================
# LINEAR COMBINATIONS AND SPAN
# ===============================================

print("=" * 50)
print("LINEAR COMBINATIONS AND SPAN")
print("=" * 50)

# Standard basis vectors in 2D
e1 = np.array([1, 0])  # i-hat
e2 = np.array([0, 1])  # j-hat

# Any 2D vector is a linear combination of e1 and e2
target = np.array([3, 2])
c1, c2 = 3, 2

result = c1 * e1 + c2 * e2
print(f"\ne1 (î) = {e1}")
print(f"e2 (ĵ) = {e2}")
print(f"\n{c1} × e1 + {c2} × e2 = {result}")
print(f"This equals our target vector: {target}")

# Visualize span of two non-collinear vectors
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

v1 = np.array([1, 0])
v2 = np.array([0, 1])

# Generate random linear combinations
np.random.seed(42)
n_points = 200
coeffs = np.random.uniform(-3, 3, (n_points, 2))
points = coeffs @ np.array([v1, v2])

ax = axes[0]
ax.scatter(points[:, 0], points[:, 1], c='steelblue', alpha=0.4, s=20)
ax.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1,
          color='red', linewidth=3, label=f'v1 = {v1}', zorder=5)
ax.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1,
          color='green', linewidth=3, label=f'v2 = {v2}', zorder=5)
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title('Span of {v1, v2} = All of R²')
ax.legend()

# Collinear vectors (span is only a line)
v3 = np.array([1, 2])
v4 = np.array([2, 4])  # Same direction as v3!

coeffs2 = np.random.uniform(-3, 3, (n_points, 2))
points2 = coeffs2 @ np.array([v3, v4])

ax = axes[1]
ax.scatter(points2[:, 0], points2[:, 1], c='coral', alpha=0.4, s=20)
ax.quiver(0, 0, v3[0], v3[1], angles='xy', scale_units='xy', scale=1,
          color='red', linewidth=3, label=f'v3 = {v3}', zorder=5)
ax.quiver(0, 0, v4[0], v4[1], angles='xy', scale_units='xy', scale=1,
          color='green', linewidth=3, label=f'v4 = {v4} (= 2×v3)', zorder=5)
ax.set_xlim(-10, 10)
ax.set_ylim(-15, 15)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title('Span of {v3, v4} = Just a line (linearly dependent!)')
ax.legend()

plt.tight_layout()
plt.show()

print("\nKey Insight: Linearly dependent vectors don't span the full space!")
print("v3 and v4 are linearly dependent because v4 = 2 × v3")
```

## Summary

| Concept | Key Formula | Purpose in Data Science |
|---------|------------|------------------------|
| **Vector** | $\vec{v} = [v_1, v_2, \ldots, v_n]$ | Represent data points, features |
| **Addition** | $\vec{v} + \vec{w}$ (component-wise) | Combining effects |
| **Scalar multiplication** | $c \cdot \vec{v}$ | Scaling, learning rates |
| **Magnitude** | $\|\vec{v}\| = \sqrt{\sum v_i^2}$ | Distance, regularization |
| **Dot product** | $\vec{v} \cdot \vec{w} = \sum v_i w_i$ | Similarity, projections |
| **Cross product** | $\vec{v} \times \vec{w}$ (3D only) | Normal vectors, areas |
| **Linear independence** | Vectors that don't reduce to others | Feature selection, rank |
