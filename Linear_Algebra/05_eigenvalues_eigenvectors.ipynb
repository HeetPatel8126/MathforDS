# **5. Eigenvalues and Eigenvectors**

## What are Eigenvalues and Eigenvectors?
An **eigenvector** of a matrix $A$ is a non-zero vector $\vec{v}$ that, when transformed by $A$, only gets **scaled** (not rotated):

$$A\vec{v} = \lambda\vec{v}$$

where:
- $\vec{v}$ is the **eigenvector** (direction that doesn't change)
- $\lambda$ is the **eigenvalue** (the scaling factor)

### Intuition
Most vectors change direction when multiplied by a matrix. Eigenvectors are **special** — they stay on the same line. The eigenvalue tells us by **how much** they're stretched or compressed.

| Eigenvalue | Effect on Eigenvector |
|-----------|----------------------|
| $\lambda > 1$ | Stretched |
| $0 < \lambda < 1$ | Compressed |
| $\lambda = 1$ | Unchanged |
| $\lambda = 0$ | Collapsed to zero |
| $\lambda < 0$ | Reversed direction |

### Why Eigenvectors Matter in Data Science

| Application | Role of Eigenvectors |
|-------------|---------------------|
| **PCA** | Eigenvectors of covariance matrix = principal components |
| **PageRank** | Eigenvector of link matrix = page importance |
| **Spectral Clustering** | Eigenvectors of Laplacian = cluster structure |
| **Stability Analysis** | Eigenvalues determine system stability |

```python
# ===============================================
# COMPUTING EIGENVALUES AND EIGENVECTORS
# ===============================================

import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import eig, det

print("=" * 50)
print("EIGENVALUES AND EIGENVECTORS")
print("=" * 50)

# Simple 2x2 matrix
A = np.array([[4, 2],
              [1, 3]])

eigenvalues, eigenvectors = eig(A)

print(f"\nMatrix A = ")
print(A)

print(f"\nEigenvalues: {eigenvalues}")
print(f"  λ₁ = {eigenvalues[0]:.4f}")
print(f"  λ₂ = {eigenvalues[1]:.4f}")

print(f"\nEigenvectors (as columns):")
print(eigenvectors)
print(f"  v₁ = [{eigenvectors[0,0]:.4f}, {eigenvectors[1,0]:.4f}]")
print(f"  v₂ = [{eigenvectors[0,1]:.4f}, {eigenvectors[1,1]:.4f}]")

# Verify: A @ v = λ * v
print("\n" + "-" * 50)
print("VERIFICATION: Av = λv")
print("-" * 50)

for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    Av = A @ v
    lam_v = lam * v
    print(f"\n  λ_{i+1} = {lam:.4f}, v_{i+1} = [{v[0]:.4f}, {v[1]:.4f}]")
    print(f"  A @ v_{i+1}  = [{Av[0]:.4f}, {Av[1]:.4f}]")
    print(f"  λ_{i+1} * v_{i+1} = [{lam_v[0]:.4f}, {lam_v[1]:.4f}]")
    print(f"  Equal? {np.allclose(Av, lam_v)}")
```

```python
# ===============================================
# VISUALIZING EIGENVECTORS
# ===============================================

print("=" * 50)
print("VISUALIZING EIGENVECTORS")
print("=" * 50)

A = np.array([[2, 1],
              [1, 2]])

eigenvalues, eigenvectors = eig(A)

print(f"\nA = \n{A}")
print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors: ")
for i in range(2):
    print(f"  v{i+1} = {eigenvectors[:, i]} (λ = {eigenvalues[i]:.1f})")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot 1: Eigenvectors stay on their line
ax = axes[0]

# Show some random vectors and their transformations
np.random.seed(42)
for _ in range(8):
    v = np.random.randn(2)
    v = v / np.linalg.norm(v) * 1.5
    Av = A @ v
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
              color='gray', alpha=0.3, linewidth=1)
    ax.quiver(0, 0, Av[0], Av[1], angles='xy', scale_units='xy', scale=1,
              color='gray', alpha=0.3, linewidth=1, linestyle='dashed')

# Show eigenvectors: these DON'T change direction!
colors = ['steelblue', 'coral']
for i in range(2):
    v = eigenvectors[:, i] * 1.5
    Av = A @ v
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
              color=colors[i], linewidth=3,
              label=f'v{i+1} (original, λ={eigenvalues[i]:.1f})')
    ax.quiver(0, 0, Av[0], Av[1], angles='xy', scale_units='xy', scale=1,
              color=colors[i], linewidth=2, linestyle='dashed', alpha=0.6,
              label=f'Av{i+1} (transformed)')

ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_title('Eigenvectors stay on their line!\n(solid = original, dashed = transformed)')
ax.legend(fontsize=8, loc='upper left')

# Plot 2: Show all vectors colored by how much their direction changes
ax = axes[1]

# Create a circle of unit vectors
angles = np.linspace(0, 2*np.pi, 100)
unit_vectors = np.array([np.cos(angles), np.sin(angles)])

# Transform them
transformed = A @ unit_vectors

# Plot original circle and transformed ellipse
ax.plot(unit_vectors[0], unit_vectors[1], 'steelblue', linewidth=2,
        label='Unit circle (before)')
ax.plot(transformed[0], transformed[1], 'coral', linewidth=2,
        label='After transformation')

# Draw eigenvectors
for i in range(2):
    v = eigenvectors[:, i]
    ax.quiver(0, 0, v[0]*eigenvalues[i], v[1]*eigenvalues[i],
              angles='xy', scale_units='xy', scale=1,
              color='green', linewidth=3, zorder=5,
              label=f'Eigenvector {i+1} (λ={eigenvalues[i]:.1f})' if i == 0 else f'Eigenvector {i+1} (λ={eigenvalues[i]:.1f})')

ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_title('Unit circle → Ellipse\n(Eigenvectors = axes of ellipse)')
ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

## Finding Eigenvalues: The Characteristic Equation

From $A\vec{v} = \lambda\vec{v}$, we rearrange:

$$(A - \lambda I)\vec{v} = \vec{0}$$

For non-trivial solutions, the matrix $(A - \lambda I)$ must be **singular**:

$$\det(A - \lambda I) = 0$$

This is the **characteristic equation**. For a 2×2 matrix:

$$\det\begin{bmatrix} a - \lambda & b \\ c & d - \lambda \end{bmatrix} = (a-\lambda)(d-\lambda) - bc = 0$$

$$\lambda^2 - (a+d)\lambda + (ad - bc) = 0$$

$$\lambda^2 - \text{tr}(A)\lambda + \det(A) = 0$$

where $\text{tr}(A) = a + d$ is the **trace** (sum of diagonal elements).

```python
# ===============================================
# CHARACTERISTIC EQUATION (BY HAND)
# ===============================================

import sympy as sp

print("=" * 50)
print("SOLVING THE CHARACTERISTIC EQUATION")
print("=" * 50)

A = np.array([[4, 2],
              [1, 3]])

print(f"\nA = \n{A}")

# By hand
trace_A = np.trace(A)
det_A = det(A)
print(f"\ntr(A) = {trace_A}")
print(f"det(A) = {det_A:.0f}")
print(f"\nCharacteristic equation: λ² - {trace_A}λ + {det_A:.0f} = 0")

# Using SymPy
lam = sp.Symbol('lambda')
A_sym = sp.Matrix([[4, 2], [1, 3]])
char_poly = A_sym.charpoly(lam)
print(f"\nSymPy characteristic polynomial: {char_poly.as_expr()}")

# Solve
eigenvals_sym = sp.solve(char_poly.as_expr(), lam)
print(f"Eigenvalues (symbolic): {eigenvals_sym}")

# Compare with NumPy
eigenvalues_np, _ = eig(A)
print(f"Eigenvalues (NumPy):    {sorted(eigenvalues_np)}")

# Key properties
print(f"\n" + "-" * 50)
print("KEY PROPERTIES")
print("-" * 50)
print(f"\nSum of eigenvalues = {sum(eigenvalues_np):.4f}")
print(f"Trace of A = {trace_A:.4f}")
print(f"Sum of eigenvalues == Trace? {np.isclose(sum(eigenvalues_np), trace_A)}")

print(f"\nProduct of eigenvalues = {np.prod(eigenvalues_np):.4f}")
print(f"Determinant of A = {det_A:.4f}")
print(f"Product of eigenvalues == Det? {np.isclose(np.prod(eigenvalues_np), det_A)}")
```

## Eigendecomposition

If $A$ has $n$ linearly independent eigenvectors, it can be **decomposed** as:

$$A = V \Lambda V^{-1}$$

where:
- $V$ = matrix whose columns are eigenvectors
- $\Lambda$ = diagonal matrix of eigenvalues

### Why is this useful?
- **Matrix powers**: $A^k = V \Lambda^k V^{-1}$ (much easier to compute!)
- **Understanding the transformation**: decompose into scale + rotate
- **PCA**: eigendecomposition of covariance matrix

```python
# ===============================================
# EIGENDECOMPOSITION
# ===============================================

print("=" * 50)
print("EIGENDECOMPOSITION: A = VΛV⁻¹")
print("=" * 50)

A = np.array([[4, 2],
              [1, 3]])

eigenvalues, V = eig(A)
Lambda = np.diag(eigenvalues)
V_inv = np.linalg.inv(V)

print(f"\nA = \n{A}")
print(f"\nV (eigenvectors as columns) = ")
print(np.round(V, 4))
print(f"\nΛ (eigenvalues on diagonal) = ")
print(np.round(Lambda, 4))

# Reconstruct A
A_reconstructed = V @ Lambda @ V_inv
print(f"\nV @ Λ @ V⁻¹ = ")
print(np.round(A_reconstructed, 10))
print(f"\nReconstruction matches A? {np.allclose(A, A_reconstructed)}")

# Power of A using eigendecomposition
print("\n" + "-" * 50)
print("MATRIX POWERS USING EIGENDECOMPOSITION")
print("-" * 50)

k = 5
A_power_direct = np.linalg.matrix_power(A, k)
A_power_eigen = V @ np.diag(eigenvalues**k) @ V_inv

print(f"\nA^{k} (direct) = ")
print(np.round(A_power_direct, 4))
print(f"\nA^{k} (via eigendecomposition: VΛ^{k}V⁻¹) = ")
print(np.round(A_power_eigen, 4))
print(f"\nMatch? {np.allclose(A_power_direct, A_power_eigen)}")
```

## Connection to PCA (Principal Component Analysis)

PCA is one of the most important applications of eigenvalues and eigenvectors in data science.

**Steps**:
1. Compute the **covariance matrix** $C$ of the data
2. Find the **eigenvectors** and **eigenvalues** of $C$
3. The eigenvectors are the **principal components** (new axes)
4. The eigenvalues tell us how much **variance** each component captures
5. Keep only the top $k$ components for dimensionality reduction

```python
# ===============================================
# PCA FROM SCRATCH USING EIGENVECTORS
# ===============================================

print("=" * 50)
print("PCA FROM SCRATCH")
print("=" * 50)

# Generate correlated 2D data
np.random.seed(42)
n = 200

# Create correlated data
mean = [0, 0]
cov = [[3, 2], [2, 2]]  # Covariance matrix
data = np.random.multivariate_normal(mean, cov, n)

# Step 1: Center the data
data_centered = data - data.mean(axis=0)

# Step 2: Compute covariance matrix
C = np.cov(data_centered.T)
print(f"\nCovariance matrix:")
print(np.round(C, 4))

# Step 3: Eigendecomposition of covariance matrix
eigenvalues, eigenvectors = eig(C)

# Sort by eigenvalue (largest first)
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print(f"\nEigenvalues (sorted): {np.round(eigenvalues, 4)}")
print(f"  PC1 captures {eigenvalues[0]/sum(eigenvalues)*100:.1f}% of variance")
print(f"  PC2 captures {eigenvalues[1]/sum(eigenvalues)*100:.1f}% of variance")

print(f"\nPrincipal Components (eigenvectors):")
print(f"  PC1 = [{eigenvectors[0,0]:.4f}, {eigenvectors[1,0]:.4f}]")
print(f"  PC2 = [{eigenvectors[0,1]:.4f}, {eigenvectors[1,1]:.4f}]")

# Visualize
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Plot 1: Original data with eigenvectors
ax = axes[0]
ax.scatter(data_centered[:, 0], data_centered[:, 1], alpha=0.4, s=20,
           color='steelblue')

# Draw eigenvectors (scaled by eigenvalue for visibility)
for i in range(2):
    v = eigenvectors[:, i] * eigenvalues[i] * 1.5
    color = 'red' if i == 0 else 'green'
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
              color=color, linewidth=3, zorder=5,
              label=f'PC{i+1} (λ={eigenvalues[i]:.2f})')

ax.set_xlim(-6, 6)
ax.set_ylim(-6, 6)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title('Original Data with\nPrincipal Components')
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.legend()

# Plot 2: Data projected onto PC1
ax = axes[1]
projected_1d = data_centered @ eigenvectors[:, 0]  # Project onto PC1
ax.hist(projected_1d, bins=30, color='coral', edgecolor='white', alpha=0.7)
ax.set_title(f'Projected onto PC1\n(captures {eigenvalues[0]/sum(eigenvalues)*100:.1f}% variance)')
ax.set_xlabel('PC1 Score')
ax.set_ylabel('Count')

# Plot 3: Data in PC space (rotated)
ax = axes[2]
data_pc = data_centered @ eigenvectors  # Project onto both PCs
ax.scatter(data_pc[:, 0], data_pc[:, 1], alpha=0.4, s=20, color='purple')
ax.set_xlim(-6, 6)
ax.set_ylim(-6, 6)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title('Data in PC Space\n(decorrelated!)')
ax.set_xlabel('PC1')
ax.set_ylabel('PC2')

plt.suptitle('PCA = Eigendecomposition of Covariance Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Verify: covariance in PC space should be diagonal
print(f"\nCovariance in original space:")
print(np.round(np.cov(data_centered.T), 4))
print(f"\nCovariance in PC space (should be diagonal):")
print(np.round(np.cov(data_pc.T), 4))
print("\nThe off-diagonal elements ≈ 0 → The data is decorrelated!")
```

## Summary

| Concept | Key Formula | Data Science Application |
|---------|------------|-------------------------|
| **Eigenvector** | $A\vec{v} = \lambda\vec{v}$ | Directions of max variance (PCA) |
| **Eigenvalue** | Scaling factor $\lambda$ | Amount of variance captured |
| **Characteristic eq.** | $\det(A - \lambda I) = 0$ | Finding eigenvalues analytically |
| **Eigendecomposition** | $A = V\Lambda V^{-1}$ | Matrix powers, analysis |
| **trace(A)** | $= \sum \lambda_i$ | Sum of eigenvalues |
| **det(A)** | $= \prod \lambda_i$ | Product of eigenvalues |
| **PCA** | Eigen-decompose covariance matrix | Dimensionality reduction |

### The Big Picture
Eigenvectors reveal the **intrinsic structure** of a linear transformation. In data science, they help us find the most **meaningful directions** in our data, enabling dimensionality reduction, feature extraction, and understanding of complex systems.
