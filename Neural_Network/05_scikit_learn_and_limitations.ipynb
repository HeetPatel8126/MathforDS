# Neural Networks with scikit-learn & Limitations

Building a neural network from scratch is great for learning, but in practice we use libraries.

**scikit-learn** has `MLPClassifier` (Multi-Layer Perceptron Classifier) — a neural network for classification.

For more advanced deep learning, consider:
- **PyTorch** (currently most popular)
- **TensorFlow / Keras**

---

## Using scikit-learn's MLPClassifier

Same RGB background color problem, but in just a few lines:

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier

# Load data
df = pd.read_csv('https://bit.ly/3GsNzGt', delimiter=",")

# Inputs: RGB scaled to 0-1
X = df.values[:, :-1] / 255.0

# Output: light (1) or dark (0) font
Y = df.values[:, -1]

# Train/test split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=1/3)
```

```python
# Create and train neural network
nn = MLPClassifier(
    solver='sgd',              # stochastic gradient descent
    hidden_layer_sizes=(3,),   # 1 hidden layer with 3 nodes
    activation='relu',         # ReLU for hidden layer
    max_iter=100_000,          # training iterations
    learning_rate_init=0.05    # learning rate
)

nn.fit(X_train, Y_train)

# Results
print(f"Training accuracy: {nn.score(X_train, Y_train):.4f}")
print(f"Testing accuracy:  {nn.score(X_test, Y_test):.4f}")
```

```python
# Inspect the learned weights and biases
print("Learned weights (W):")
for i, coef in enumerate(nn.coefs_):
    print(f"  Layer {i+1}: shape {coef.shape}")
    print(f"  {coef}\n")

print("Learned biases (B):")
for i, intercept in enumerate(nn.intercepts_):
    print(f"  Layer {i+1}: {intercept}")
```

## Comparison: From Scratch vs scikit-learn

| Aspect | From Scratch (NumPy) | scikit-learn |
| --- | --- | --- |
| Lines of code | ~50 lines | ~10 lines |
| Accuracy | 97–99% | ~99%+ |
| Learning value | High (understand internals) | Low (black box) |
| Flexibility | Full control | Limited options |
| Production use | No | Somewhat |

**Verdict:** Build from scratch to *learn*, use libraries to *ship*.

## Limitations of Neural Networks

### 1. Overfitting
Neural networks are *too flexible* — they can memorize training data instead of learning general patterns.

**Example:** Andrew Ng shared that a radiology AI trained at Stanford Hospital performed well on Stanford's data but **degraded significantly** when tested at a different hospital with different equipment.

### 2. Data Hungry
- Need **huge amounts** of labeled data
- Collecting real-world data is expensive and hard
- Autonomous vehicles need training on countless edge cases: weather, lighting, costumes, graffiti, etc.

### 3. Narrow Intelligence
- Neural networks solve **one specific problem** — they can't reason broadly
- An AI that plays Chess brilliantly can't do anything else
- Performance in games ≠ general intelligence

### 4. Real World vs Simulations
- AI excels in games because unlimited labeled data can be generated in simulations
- The real world has **infinite unpredictability** and rare events
- Self-driving cars don't use end-to-end neural networks — they use neural nets as *one small piece* alongside rule-based systems and sensors

## When to Use What — Decision Guide

| Problem Type | Recommended Approach |
| --- | --- |
| Structured/tabular data | Linear/Logistic regression, Gradient boosted trees |
| Image classification | Convolutional Neural Networks (CNNs) |
| Text / NLP | Transformers, RNNs |
| Time series forecasting | LSTM, traditional statistics |
| Simple rule-based logic | Don't use ML at all — just code the rules |

> **"Use whatever tool works best — don't be partial to neural networks or any other tool."**

## Key Takeaways

1. **scikit-learn's MLPClassifier** gives you a neural network in ~10 lines of code
2. For serious deep learning, use **PyTorch** or **TensorFlow**
3. Neural networks **overfit easily** — always validate on unseen data
4. They require **massive amounts of labeled data** to work well
5. Neural networks are **narrow** — they can't reason or generalize beyond their training
6. Always ask: "Is a simpler model good enough?" — often the answer is yes
