# Forward Propagation

**Forward propagation** = running inputs *forward* through the network to get a prediction.

It's a 4-step pipeline:

```
Step 1: Z1 = W_hidden · X + B_hidden        (weighted sum in hidden layer)
Step 2: A1 = ReLU(Z1)                        (activate hidden layer)
Step 3: Z2 = W_output · A1 + B_output        (weighted sum in output layer)
Step 4: A2 = sigmoid(Z2)                     (activate output → probability)
```

### Notation Guide
| Symbol | Meaning |
| --- | --- |
| Z | Unactivated output (raw weighted sum) |
| A | Activated output (after activation function) |
| 1 or 2 | Layer number (1 = hidden, 2 = output) |
| W | Weight matrix |
| B | Bias vector |
| X | Input vector (R, G, B values) |

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# Load and prepare data
all_data = pd.read_csv("https://tinyurl.com/y2qmhfsr")
all_inputs = all_data.iloc[:, 0:3].values / 255.0  # Scale to 0-1
all_outputs = all_data.iloc[:, -1].values

# Split: 2/3 training, 1/3 testing
X_train, X_test, Y_train, Y_test = train_test_split(
    all_inputs, all_outputs, test_size=1/3
)
n = X_train.shape[0]
````

````python
# Initialize random weights and biases
w_hidden = np.random.rand(3, 3)
w_output = np.random.rand(1, 3)
b_hidden = np.random.rand(3, 1)
b_output = np.random.rand(1, 1)

# Activation functions
relu = lambda x: np.maximum(x, 0)
logistic = lambda x: 1 / (1 + np.exp(-x))
````

## The Forward Propagation Function

This is the heart of the neural network. In just 4 lines, we process an input through both layers using **matrix multiplication**:

````python
def forward_prop(X):
    Z1 = w_hidden @ X + b_hidden     # Step 1: Hidden layer weighted sum
    A1 = relu(Z1)                     # Step 2: Apply ReLU
    Z2 = w_output @ A1 + b_output    # Step 3: Output layer weighted sum
    A2 = logistic(Z2)                 # Step 4: Apply Sigmoid → probability
    return Z1, A1, Z2, A2
````

## Step-by-Step Walkthrough

Let's trace what happens when we feed in a single color (e.g., pink = RGB 255, 192, 203):

````python
# Input: pink color (scaled to 0-1)
X_pink = np.array([[255, 192, 203]]).transpose() / 255.0
print("Input X (scaled):")
print(X_pink)

# Step 1: Z1 = W_hidden @ X + B_hidden
Z1 = w_hidden @ X_pink + b_hidden
print("\nStep 1 — Z1 (hidden layer, unactivated):")
print(Z1)

# Step 2: A1 = ReLU(Z1)
A1 = relu(Z1)
print("\nStep 2 — A1 (hidden layer, after ReLU):")
print(A1)
print("(Negative values would become 0)")

# Step 3: Z2 = W_output @ A1 + B_output
Z2 = w_output @ A1 + b_output
print("\nStep 3 — Z2 (output layer, unactivated):")
print(Z2)

# Step 4: A2 = sigmoid(Z2)
A2 = logistic(Z2)
print("\nStep 4 — A2 (final prediction probability):")
print(A2)
print(f"\nPrediction: {'LIGHT font' if A2[0][0] >= 0.5 else 'DARK font'}")
````

## Calculating Accuracy (Untrained Network)

Since the weights are **random**, the accuracy will be around 55–67% (basically a coin flip for binary classification).

````python
# Run ALL test data through the network
test_predictions = forward_prop(X_test.transpose())[3]  # grab A2

# Convert probabilities to 0/1 and compare with actual labels
test_comparisons = np.equal(
    (test_predictions >= 0.5).flatten().astype(int), 
    Y_test
)
accuracy = sum(test_comparisons.astype(int) / X_test.shape[0])

print(f"Accuracy with RANDOM weights: {accuracy:.2%}")
print("\n(This is expected — we haven't trained the network yet!)")
````

## Key Takeaways

1. **Forward propagation** = pushing inputs through the layers to get an output
2. At each layer: **multiply weights → add bias → apply activation function**
3. The `@` operator does **matrix multiplication** — it handles all nodes at once
4. With random weights, accuracy is ~50–67% (basically guessing)
5. We need **backpropagation** + **gradient descent** to train the network (next notebook!)
