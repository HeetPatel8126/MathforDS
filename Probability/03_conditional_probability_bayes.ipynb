# **3. Conditional Probability & Bayes' Theorem**

## Conditional Probability
The probability of event A occurring **given that** event B has already occurred:

$$P(A|B) = \frac{P(A \cap B)}{P(B)}$$

Read as: "Probability of A given B"

### Key Insight
When we know B has occurred, we're "zooming in" on a smaller sample space.

## Why This Matters in Data Science
- **Classification**: P(Class | Features) - What class given these features?
- **Medical Diagnosis**: P(Disease | Positive Test)
- **Spam Detection**: P(Spam | Contains "lottery")
- **Recommendation Systems**: P(Like Item B | Liked Item A)

```python
# ===============================================
# CONDITIONAL PROBABILITY: BASICS
# ===============================================

from fractions import Fraction

print("=" * 50)
print("CONDITIONAL PROBABILITY")
print("=" * 50)

# Example: Drawing cards
# What's P(Heart | Red card)?

print("\nP(Heart | Red card) = ?")
print("\nGiven we drew a red card (26 cards: 13 hearts + 13 diamonds)")
print("What's the probability it's a heart?")

# Method 1: Direct calculation
# Out of 26 red cards, 13 are hearts
P_heart_given_red = Fraction(13, 26)
print(f"\nDirect: 13 hearts / 26 red cards = {P_heart_given_red}")

# Method 2: Using formula P(A|B) = P(A∩B) / P(B)
P_heart_and_red = Fraction(13, 52)  # All hearts are red
P_red = Fraction(26, 52)

P_heart_given_red_formula = P_heart_and_red / P_red
print(f"\nUsing formula: P(Heart ∩ Red) / P(Red)")
print(f"             = {P_heart_and_red} / {P_red} = {P_heart_given_red_formula}")
```

```python
# ===============================================
# CONDITIONAL PROBABILITY: DATA SCIENCE EXAMPLE
# ===============================================

from fractions import Fraction
import pandas as pd

print("=" * 50)
print("REAL-WORLD EXAMPLE: Customer Purchases")
print("=" * 50)

# Simulated data: 100 customers
# Bought Premium | Clicked Ad = 30
# Bought Premium | No Click = 10
# No Premium | Clicked Ad = 20
# No Premium | No Click = 40

data = {
    'Clicked Ad': [30, 20, 50],
    'No Click': [10, 40, 50],
    'Total': [40, 60, 100]
}
df = pd.DataFrame(data, index=['Bought Premium', 'No Premium', 'Total'])
print("\nContingency Table:")
print(df)

# P(Premium | Clicked Ad)
P_premium_given_click = Fraction(30, 50)
print(f"\nP(Premium | Clicked Ad) = 30/50 = {P_premium_given_click} = {float(P_premium_given_click):.1%}")

# P(Premium | No Click)
P_premium_given_no_click = Fraction(10, 50)
print(f"P(Premium | No Click) = 10/50 = {P_premium_given_no_click} = {float(P_premium_given_no_click):.1%}")

# Overall P(Premium)
P_premium = Fraction(40, 100)
print(f"\nOverall P(Premium) = 40/100 = {P_premium} = {float(P_premium):.1%}")

print("\n📊 Insight: Clicking the ad INCREASES purchase probability!")
print(f"   From {float(P_premium_given_no_click):.1%} to {float(P_premium_given_click):.1%}")
```

# **Bayes' Theorem**

## The Formula
$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$

### Components
| Term | Name | Description |
|------|------|-------------|
| $P(A|B)$ | **Posterior** | Updated probability after seeing evidence |
| $P(B|A)$ | **Likelihood** | Probability of evidence given hypothesis |
| $P(A)$ | **Prior** | Initial probability before evidence |
| $P(B)$ | **Evidence** | Total probability of observing the evidence |

### Expanded Form (Law of Total Probability)
$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B|A) \cdot P(A) + P(B|A') \cdot P(A')}$$

## Why Bayes' Theorem is Crucial
- Allows us to "flip" conditional probabilities
- We often know P(Data | Hypothesis) but want P(Hypothesis | Data)
- Foundation of Bayesian statistics and Naive Bayes classifier

```python
# ===============================================
# BAYES' THEOREM: MEDICAL TESTING EXAMPLE
# ===============================================

print("=" * 50)
print("BAYES' THEOREM: Medical Test Example")
print("=" * 50)

# Classic example: Does a positive test mean you have the disease?

# Given:
P_disease = 0.01       # 1% of population has the disease (Prior)
P_positive_given_disease = 0.99  # Test sensitivity: 99% (Likelihood)
P_positive_given_no_disease = 0.05  # False positive rate: 5%

print("\nGiven Information:")
print(f"• P(Disease) = {P_disease:.1%} (1 in 100 people have it)")
print(f"• P(+Test | Disease) = {P_positive_given_disease:.1%} (sensitivity)")
print(f"• P(+Test | No Disease) = {P_positive_given_no_disease:.1%} (false positive)")

# Calculate P(Positive Test) using Total Probability
P_no_disease = 1 - P_disease
P_positive = (P_positive_given_disease * P_disease + 
              P_positive_given_no_disease * P_no_disease)

print(f"\nP(+Test) = P(+|D)·P(D) + P(+|¬D)·P(¬D)")
print(f"        = {P_positive_given_disease}×{P_disease} + {P_positive_given_no_disease}×{P_no_disease}")
print(f"        = {P_positive:.4f}")

# Apply Bayes' Theorem
P_disease_given_positive = (P_positive_given_disease * P_disease) / P_positive

print(f"\n\n🔍 Bayes' Theorem: P(Disease | +Test) = ?")
print(f"\nP(D|+) = P(+|D) × P(D) / P(+)")
print(f"       = {P_positive_given_disease} × {P_disease} / {P_positive:.4f}")
print(f"       = {P_disease_given_positive:.4f}")
print(f"       ≈ {P_disease_given_positive:.1%}")

print("\n" + "=" * 50)
print("⚠️  COUNTERINTUITIVE RESULT!")
print(f"Even with a positive test, there's only ~{P_disease_given_positive:.1%} chance")
print("of actually having the disease!")
print("\nThis is because the disease is RARE (low prior).")
print("Most positive tests are FALSE POSITIVES.")
print("=" * 50)
```

```python
# ===============================================
# VISUALIZING BAYES' THEOREM
# ===============================================

import matplotlib.pyplot as plt
import numpy as np

# Population of 10,000 people
N = 10000

# Using our medical test example
have_disease = int(N * 0.01)  # 100 people
no_disease = N - have_disease  # 9900 people

# True positives and false positives
true_positives = int(have_disease * 0.99)  # 99
false_positives = int(no_disease * 0.05)   # 495

total_positive = true_positives + false_positives

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Population breakdown
ax1 = axes[0]
sizes = [have_disease, no_disease]
labels = [f'Has Disease\n({have_disease})', f'No Disease\n({no_disease})']
colors = ['#ff6b6b', '#4ecdc4']
ax1.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', 
        startangle=90, explode=(0.05, 0))
ax1.set_title('Population Distribution\n(10,000 people)')

# Plot 2: Among positive tests
ax2 = axes[1]
sizes2 = [true_positives, false_positives]
labels2 = [f'True Positives\n(Actually Sick: {true_positives})', 
           f'False Positives\n(Healthy: {false_positives})']
colors2 = ['#ff6b6b', '#ffe66d']
wedges, texts, autotexts = ax2.pie(sizes2, labels=labels2, colors=colors2, 
                                    autopct='%1.1f%%', startangle=90,
                                    explode=(0.05, 0))
ax2.set_title(f'Among ALL Positive Tests\n({total_positive} people)')

plt.tight_layout()
plt.show()

print(f"\nOf {total_positive} positive tests:")
print(f"  • Only {true_positives} truly have the disease")
print(f"  • {false_positives} are false alarms!")
print(f"\nP(Disease | Positive) = {true_positives}/{total_positive} = {true_positives/total_positive:.1%}")
```

```python
# ===============================================
# BAYES' THEOREM: SPAM CLASSIFICATION
# ===============================================

print("=" * 50)
print("BAYES' THEOREM: Spam Email Detection")
print("=" * 50)

# Email spam detection example
# Given: Email contains the word "FREE"
# What's the probability it's spam?

# Prior probabilities (from training data)
P_spam = 0.30        # 30% of emails are spam
P_not_spam = 0.70    # 70% are legitimate

# Likelihoods
P_free_given_spam = 0.65      # 65% of spam contains "FREE"
P_free_given_not_spam = 0.05  # 5% of legit emails contain "FREE"

print("\nPrior knowledge:")
print(f"  P(Spam) = {P_spam:.0%}")
print(f"  P('FREE' | Spam) = {P_free_given_spam:.0%}")
print(f"  P('FREE' | Not Spam) = {P_free_given_not_spam:.0%}")

# Calculate P("FREE")
P_free = P_free_given_spam * P_spam + P_free_given_not_spam * P_not_spam

# Bayes' Theorem
P_spam_given_free = (P_free_given_spam * P_spam) / P_free

print(f"\n📧 If an email contains 'FREE':")
print(f"   P(Spam | 'FREE') = {P_spam_given_free:.1%}")

# Now add another word: "WINNER"
print("\n" + "-" * 50)
print("Adding more evidence: Email also contains 'WINNER'")

# Update prior with what we've learned
P_spam_updated = P_spam_given_free
P_not_spam_updated = 1 - P_spam_updated

# Likelihoods for "WINNER"
P_winner_given_spam = 0.45
P_winner_given_not_spam = 0.02

P_winner = (P_winner_given_spam * P_spam_updated + 
            P_winner_given_not_spam * P_not_spam_updated)

P_spam_given_both = (P_winner_given_spam * P_spam_updated) / P_winner

print(f"   P(Spam | 'FREE' AND 'WINNER') = {P_spam_given_both:.1%}")
print("\n🎯 This is how Naive Bayes classifier works!")
```

## Summary

| Concept | Formula | Key Insight |
|---------|---------|-------------|
| **Conditional P** | $P(A|B) = \frac{P(A \cap B)}{P(B)}$ | Probability given new information |
| **Bayes' Theorem** | $P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$ | "Flip" conditional probabilities |
| **Prior** | $P(A)$ | Belief before evidence |
| **Posterior** | $P(A|B)$ | Updated belief after evidence |
| **Likelihood** | $P(B|A)$ | How likely is evidence given hypothesis |
