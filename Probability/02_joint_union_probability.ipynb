# **2. Joint and Union Probabilities**

## Joint Probability (Intersection)
The probability that **both** events A AND B occur together.

$$P(A \cap B) = P(A \text{ and } B)$$

### For Independent Events
Two events are **independent** if one doesn't affect the other:
$$P(A \cap B) = P(A) \cdot P(B)$$

### For Dependent Events
$$P(A \cap B) = P(A) \cdot P(B|A)$$

where $P(B|A)$ is the probability of B given A has occurred.

## Union Probability (OR)
The probability that event A OR event B (or both) occur:

$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

We subtract $P(A \cap B)$ to avoid counting the overlap twice.

```python
# ===============================================
# JOINT PROBABILITY: INDEPENDENT EVENTS
# ===============================================

from fractions import Fraction

print("=" * 50)
print("JOINT PROBABILITY: Independent Events")
print("=" * 50)

# Example: Rolling two dice
# P(First die = 6 AND Second die = 6)

P_six = Fraction(1, 6)
print(f"\nP(Die 1 = 6) = {P_six}")
print(f"P(Die 2 = 6) = {P_six}")

# Since dice rolls are independent
P_both_six = P_six * P_six
print(f"\nP(Both dice = 6) = P(D1=6) × P(D2=6)")
print(f"                 = {P_six} × {P_six} = {P_both_six}")
print(f"                 ≈ {float(P_both_six):.4f} or about 2.78%")

# Another example: Coin flip AND die roll
print("\n" + "-" * 50)
print("P(Heads AND Roll > 4):")
P_heads = Fraction(1, 2)
P_greater_than_4 = Fraction(2, 6)  # 5 or 6
P_joint = P_heads * P_greater_than_4
print(f"P(Heads) = {P_heads}")
print(f"P(Roll > 4) = {P_greater_than_4}")
print(f"P(Heads AND Roll > 4) = {P_joint} ≈ {float(P_joint):.4f}")
```

```python
# ===============================================
# JOINT PROBABILITY: DEPENDENT EVENTS
# ===============================================

from fractions import Fraction

print("=" * 50)
print("JOINT PROBABILITY: Dependent Events")
print("=" * 50)

# Example: Drawing cards WITHOUT replacement
# P(Both cards are Aces)

print("\nDrawing 2 cards WITHOUT replacement:")

# First card is an Ace
P_first_ace = Fraction(4, 52)
print(f"P(1st card is Ace) = 4/52 = {P_first_ace}")

# Second card is an Ace GIVEN first was an Ace
P_second_ace_given_first = Fraction(3, 51)  # Only 3 aces left out of 51 cards
print(f"P(2nd Ace | 1st Ace) = 3/51 = {P_second_ace_given_first}")

# Joint probability
P_both_aces = P_first_ace * P_second_ace_given_first
print(f"\nP(Both Aces) = P(1st Ace) × P(2nd Ace | 1st Ace)")
print(f"             = {P_first_ace} × {P_second_ace_given_first}")
print(f"             = {P_both_aces} ≈ {float(P_both_aces):.4f}")

# Compare with independent case (WITH replacement)
print("\n" + "-" * 50)
print("With replacement (independent):")
P_both_aces_ind = P_first_ace * P_first_ace
print(f"P(Both Aces) = {P_first_ace} × {P_first_ace} = {P_both_aces_ind}")
print(f"             ≈ {float(P_both_aces_ind):.4f}")
```

```python
# ===============================================
# UNION PROBABILITY
# ===============================================

from fractions import Fraction

print("=" * 50)
print("UNION PROBABILITY: Addition Rule")
print("=" * 50)

# Example 1: Overlapping events
print("\nExample 1: P(King OR Heart) from a deck of cards")

P_king = Fraction(4, 52)
P_heart = Fraction(13, 52)
P_king_and_heart = Fraction(1, 52)  # King of Hearts

# Union formula: P(A ∪ B) = P(A) + P(B) - P(A ∩ B)
P_king_or_heart = P_king + P_heart - P_king_and_heart

print(f"P(King) = {P_king}")
print(f"P(Heart) = {P_heart}")
print(f"P(King AND Heart) = {P_king_and_heart}")
print(f"\nP(King OR Heart) = P(K) + P(H) - P(K∩H)")
print(f"                 = {P_king} + {P_heart} - {P_king_and_heart}")
print(f"                 = {P_king_or_heart} ≈ {float(P_king_or_heart):.4f}")

# Verify: Count actual outcomes
# 4 Kings + 13 Hearts - 1 King of Hearts = 16 cards
print(f"\nVerification: 4 Kings + 13 Hearts - 1 King♥ = 16 favorable cards")
print(f"              16/52 = {Fraction(16, 52)}")
```

```python
# ===============================================
# MUTUALLY EXCLUSIVE EVENTS
# ===============================================

from fractions import Fraction

print("=" * 50)
print("MUTUALLY EXCLUSIVE EVENTS")
print("=" * 50)

# Mutually exclusive: Events that CANNOT occur together
# P(A ∩ B) = 0, so P(A ∪ B) = P(A) + P(B)

print("\nExample: Rolling a die")
print("Events: A = 'Roll a 1', B = 'Roll a 6'")
print("These are MUTUALLY EXCLUSIVE (can't happen together)")

P_roll_1 = Fraction(1, 6)
P_roll_6 = Fraction(1, 6)

# Since mutually exclusive
P_roll_1_or_6 = P_roll_1 + P_roll_6
print(f"\nP(1) = {P_roll_1}")
print(f"P(6) = {P_roll_6}")
print(f"P(1 OR 6) = P(1) + P(6) = {P_roll_1_or_6}")

# Another example
print("\n" + "-" * 50)
print("P(Roll odd OR Roll even):")
P_odd = Fraction(3, 6)
P_even = Fraction(3, 6)
P_odd_or_even = P_odd + P_even
print(f"P(Odd) + P(Even) = {P_odd} + {P_even} = {P_odd_or_even}")
print("(They cover the entire sample space!)")
```

```python
# ===============================================
# VISUALIZING JOINT VS UNION PROBABILITY
# ===============================================

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Rectangle

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Simulation data for a grid (like rolling two dice)
die1 = np.arange(1, 7)
die2 = np.arange(1, 7)
D1, D2 = np.meshgrid(die1, die2)

# Plot 1: Joint Probability - Both dice = 6
ax1 = axes[0]
ax1.set_xlim(0.5, 6.5)
ax1.set_ylim(0.5, 6.5)

for i in range(1, 7):
    for j in range(1, 7):
        color = 'green' if (i == 6 and j == 6) else 'lightgray'
        rect = Rectangle((i-0.4, j-0.4), 0.8, 0.8, fill=True, 
                         facecolor=color, edgecolor='black')
        ax1.add_patch(rect)
        ax1.text(i, j, f'{i},{j}', ha='center', va='center', fontsize=8)

ax1.set_xlabel('Die 1')
ax1.set_ylabel('Die 2')
ax1.set_title('Joint Probability\nP(Die1=6 AND Die2=6) = 1/36')
ax1.set_xticks(range(1, 7))
ax1.set_yticks(range(1, 7))
ax1.set_aspect('equal')

# Plot 2: Union Probability - Die1=6 OR Die2=6
ax2 = axes[1]
ax2.set_xlim(0.5, 6.5)
ax2.set_ylim(0.5, 6.5)

for i in range(1, 7):
    for j in range(1, 7):
        color = 'blue' if (i == 6 or j == 6) else 'lightgray'
        if i == 6 and j == 6:
            color = 'green'  # Intersection
        rect = Rectangle((i-0.4, j-0.4), 0.8, 0.8, fill=True, 
                         facecolor=color, edgecolor='black', alpha=0.7)
        ax2.add_patch(rect)
        ax2.text(i, j, f'{i},{j}', ha='center', va='center', fontsize=8)

ax2.set_xlabel('Die 1')
ax2.set_ylabel('Die 2')
ax2.set_title('Union Probability\nP(Die1=6 OR Die2=6) = 11/36')
ax2.set_xticks(range(1, 7))
ax2.set_yticks(range(1, 7))
ax2.set_aspect('equal')

plt.tight_layout()
plt.show()

# Calculate
print("\nCalculation check:")
print("P(D1=6) = 6/36, P(D2=6) = 6/36, P(Both=6) = 1/36")
print("P(D1=6 OR D2=6) = 6/36 + 6/36 - 1/36 = 11/36")
```

---
## Summary

| Concept | Formula | When to Use |
|---------|---------|-------------|
| **Joint (Independent)** | $P(A \cap B) = P(A) \cdot P(B)$ | Events don't affect each other |
| **Joint (Dependent)** | $P(A \cap B) = P(A) \cdot P(B|A)$ | One event affects the other |
| **Union (General)** | $P(A \cup B) = P(A) + P(B) - P(A \cap B)$ | Either or both events |
| **Union (Mutually Exclusive)** | $P(A \cup B) = P(A) + P(B)$ | Events can't occur together |
