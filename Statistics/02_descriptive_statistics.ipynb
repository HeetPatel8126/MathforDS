# **2. Descriptive Statistics**

Descriptive statistics **summarize** and **describe** the main features of a dataset.

## Measures of Central Tendency

| Measure | Formula | When to Use |
|---------|---------|-------------|
| **Mean** | $\bar{x} = \frac{\sum x_i}{n}$ | Symmetric data, no outliers |
| **Median** | Middle value | Skewed data, outliers present |
| **Mode** | Most frequent value | Categorical data, multimodal |

## Measures of Spread

| Measure | Formula | Interpretation |
|---------|---------|---------------|
| **Range** | max - min | Simplest spread measure |
| **Variance** | $s^2 = \frac{\sum(x_i - \bar{x})^2}{n-1}$ | Average squared deviation |
| **Std Deviation** | $s = \sqrt{s^2}$ | Spread in original units |
| **IQR** | Q3 - Q1 | Robust to outliers |

````python
# ===============================================
# MEASURES OF CENTRAL TENDENCY
# ===============================================

import numpy as np
from scipy import stats

print("=" * 50)
print("MEASURES OF CENTRAL TENDENCY")
print("=" * 50)

# Example dataset: Salaries (in thousands)
salaries = np.array([45, 52, 48, 55, 62, 58, 51, 49, 53, 250])  # Note the outlier!

print(f"\nDataset: {salaries}")
print(f"(Notice the outlier: 250)")

# Mean
mean = np.mean(salaries)
print(f"\nMean (x̄): ${mean:.1f}K")

# Median
median = np.median(salaries)
print(f"Median: ${median:.1f}K")

# Mode (using scipy)
mode_result = stats.mode(salaries, keepdims=True)
print(f"Mode: ${mode_result.mode[0]}K (count: {mode_result.count[0]})")

# Without outlier
print("\n--- Without outlier ---")
salaries_clean = salaries[salaries < 100]
print(f"Mean: ${np.mean(salaries_clean):.1f}K")
print(f"Median: ${np.median(salaries_clean):.1f}K")

print("\n💡 Notice: Median is more robust to outliers!")
````

````python
# ===============================================
# WEIGHTED MEAN
# ===============================================

import numpy as np

print("=" * 50)
print("WEIGHTED MEAN")
print("=" * 50)

# Example: Calculate GPA
# Grades and their credit hours
grades = np.array([4.0, 3.5, 3.0, 4.0, 3.5])  # Grade points
credits = np.array([3, 4, 3, 2, 3])           # Credit hours

print(f"\nCourses:")
for i, (g, c) in enumerate(zip(grades, credits), 1):
    print(f"  Course {i}: Grade = {g}, Credits = {c}")

# Simple mean (wrong for GPA)
simple_mean = np.mean(grades)
print(f"\nSimple Mean: {simple_mean:.2f}")

# Weighted mean (correct)
weighted_mean = np.average(grades, weights=credits)
print(f"Weighted Mean (GPA): {weighted_mean:.2f}")

# Manual calculation
manual = np.sum(grades * credits) / np.sum(credits)
print(f"\nManual: Σ(grade × credits) / Σ(credits)")
print(f"      = {np.sum(grades * credits)} / {np.sum(credits)} = {manual:.2f}")
````

````python
# ===============================================
# MEASURES OF SPREAD
# ===============================================

import numpy as np

print("=" * 50)
print("MEASURES OF SPREAD")
print("=" * 50)

data = np.array([23, 25, 28, 30, 32, 35, 38, 40, 45, 50])

print(f"\nDataset: {data}")
print(f"Mean: {np.mean(data):.2f}")

# Range
range_val = np.max(data) - np.min(data)
print(f"\n1. Range: {np.max(data)} - {np.min(data)} = {range_val}")

# Variance (sample variance uses n-1)
variance = np.var(data, ddof=1)  # ddof=1 for sample variance
print(f"\n2. Variance (s²): {variance:.2f}")

# Standard Deviation
std_dev = np.std(data, ddof=1)
print(f"\n3. Standard Deviation (s): {std_dev:.2f}")

# Manual variance calculation
mean = np.mean(data)
squared_diffs = (data - mean) ** 2
manual_var = np.sum(squared_diffs) / (len(data) - 1)
print(f"\nManual calculation:")
print(f"  s² = Σ(xᵢ - x̄)² / (n-1)")
print(f"     = {np.sum(squared_diffs):.2f} / {len(data)-1} = {manual_var:.2f}")

# IQR
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1
print(f"\n4. IQR: Q3 - Q1 = {q3} - {q1} = {iqr}")
````

````python
# ===============================================
# VISUALIZING DESCRIPTIVE STATISTICS
# ===============================================

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Create three different distributions
symmetric = np.random.normal(50, 10, 1000)
right_skewed = np.random.exponential(10, 1000) + 30
left_skewed = 100 - np.random.exponential(10, 1000)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, data, title in zip(axes, 
                           [symmetric, right_skewed, left_skewed],
                           ['Symmetric', 'Right-Skewed', 'Left-Skewed']):
    ax.hist(data, bins=30, color='steelblue', edgecolor='white', alpha=0.7)
    
    mean = np.mean(data)
    median = np.median(data)
    
    ax.axvline(mean, color='red', linewidth=2, label=f'Mean = {mean:.1f}')
    ax.axvline(median, color='green', linewidth=2, linestyle='--', label=f'Median = {median:.1f}')
    
    ax.set_title(title)
    ax.legend()

plt.suptitle('Mean vs Median in Different Distributions', fontsize=14)
plt.tight_layout()
plt.show()

print("💡 Key Insight:")
print("  - Symmetric: Mean ≈ Median")
print("  - Right-Skewed: Mean > Median (outliers pull mean up)")
print("  - Left-Skewed: Mean < Median (outliers pull mean down)")
````

````python
# ===============================================
# BOX PLOT: 5-NUMBER SUMMARY
# ===============================================

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Create sample data with outliers
data = np.concatenate([np.random.normal(50, 10, 100), [10, 95, 100]])

# Calculate 5-number summary
minimum = np.min(data)
q1 = np.percentile(data, 25)
median = np.median(data)
q3 = np.percentile(data, 75)
maximum = np.max(data)
iqr = q3 - q1

print("=" * 50)
print("5-NUMBER SUMMARY")
print("=" * 50)
print(f"\nMinimum: {minimum:.1f}")
print(f"Q1 (25th percentile): {q1:.1f}")
print(f"Median (Q2): {median:.1f}")
print(f"Q3 (75th percentile): {q3:.1f}")
print(f"Maximum: {maximum:.1f}")
print(f"IQR: {iqr:.1f}")

# Outlier bounds
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
print(f"\nOutlier bounds: [{lower_bound:.1f}, {upper_bound:.1f}]")

# Box plot
fig, ax = plt.subplots(figsize=(10, 4))
bp = ax.boxplot(data, vert=False, widths=0.5)
ax.scatter(data, np.ones(len(data)) + 0.2*np.random.randn(len(data)), 
          alpha=0.4, color='steelblue')
ax.set_xlabel('Value')
ax.set_title('Box Plot with 5-Number Summary')
ax.set_yticks([1])
ax.set_yticklabels(['Data'])
plt.show()
````

---
## Summary: Choosing the Right Measure

| Data Type | Central Tendency | Spread |
|-----------|------------------|--------|
| Symmetric, no outliers | Mean | Std Dev |
| Skewed or outliers | Median | IQR |
| Categorical | Mode | — |
