# **1. Introduction to Statistics**

## What is Data?
**Data** is a collection of facts, observations, or measurements that can be analyzed to gain insights.

### Types of Data

| Type | Description | Examples |
|------|-------------|----------|
| **Quantitative (Numerical)** | Measurable quantities | Height, weight, temperature |
| **Qualitative (Categorical)** | Descriptive categories | Color, gender, country |
| **Discrete** | Countable values | Number of students, dice rolls |
| **Continuous** | Any value in a range | Time, distance, temperature |

## Descriptive vs Inferential Statistics

| Aspect | Descriptive | Inferential |
|--------|-------------|-------------|
| **Purpose** | Summarize data | Draw conclusions beyond data |
| **Scope** | Describes sample/population | Makes predictions about population |
| **Examples** | Mean, median, charts | Confidence intervals, hypothesis tests |
| **Question** | "What happened?" | "What does this mean?" |

---
# **Populations, Samples, and Bias**

## Definitions

| Term | Symbol | Definition |
|------|--------|------------|
| **Population** | N | The entire group of interest |
| **Sample** | n | A subset of the population |
| **Parameter** | μ, σ | A value describing the population |
| **Statistic** | x̄, s | A value calculated from a sample |

## Why Sample?
- Population may be infinite or inaccessible
- Sampling saves time and resources
- Sometimes measuring destroys the item (quality testing)

## Types of Bias

| Bias Type | Description | Example |
|-----------|-------------|--------|
| **Selection Bias** | Non-random sample selection | Only surveying volunteers |
| **Survivorship Bias** | Only analyzing "survivors" | Studying successful companies only |
| **Confirmation Bias** | Seeking confirming evidence | Ignoring contradictory data |
| **Response Bias** | Distorted answers | Leading survey questions |

```python
# ===============================================
# POPULATION VS SAMPLE DEMONSTRATION
# ===============================================

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Create a "population" of 10,000 people with heights (in cm)
population = np.random.normal(loc=170, scale=10, size=10000)

print("=" * 50)
print("POPULATION VS SAMPLE")
print("=" * 50)

# True population parameters (usually unknown!)
pop_mean = np.mean(population)
pop_std = np.std(population)
print(f"\nPopulation Parameters (usually unknown):")
print(f"  μ (mean) = {pop_mean:.2f} cm")
print(f"  σ (std)  = {pop_std:.2f} cm")

# Take a random sample
sample = np.random.choice(population, size=50, replace=False)
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)  # ddof=1 for sample std

print(f"\nSample Statistics (n=50):")
print(f"  x̄ (mean) = {sample_mean:.2f} cm")
print(f"  s (std)  = {sample_std:.2f} cm")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(population, bins=50, color='steelblue', edgecolor='white', alpha=0.7)
axes[0].axvline(pop_mean, color='red', linewidth=2, label=f'μ = {pop_mean:.1f}')
axes[0].set_title(f'Population (N = {len(population):,})')
axes[0].set_xlabel('Height (cm)')
axes[0].legend()

axes[1].hist(sample, bins=15, color='coral', edgecolor='white', alpha=0.7)
axes[1].axvline(sample_mean, color='red', linewidth=2, label=f'x̄ = {sample_mean:.1f}')
axes[1].set_title(f'Sample (n = {len(sample)})')
axes[1].set_xlabel('Height (cm)')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# ===============================================
# SAMPLING VARIABILITY
# ===============================================

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Same population
population = np.random.normal(loc=170, scale=10, size=10000)
pop_mean = np.mean(population)

# Take multiple samples and compute their means
sample_means = []
n_samples = 1000
sample_size = 30

for _ in range(n_samples):
    sample = np.random.choice(population, size=sample_size, replace=False)
    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)

print("=" * 50)
print(f"SAMPLING VARIABILITY ({n_samples} samples of size {sample_size})")
print("=" * 50)
print(f"\nPopulation mean: {pop_mean:.2f}")
print(f"Mean of sample means: {np.mean(sample_means):.2f}")
print(f"Std of sample means: {np.std(sample_means):.2f}")
print(f"Theoretical SE (σ/√n): {np.std(population)/np.sqrt(sample_size):.2f}")

plt.figure(figsize=(10, 5))
plt.hist(sample_means, bins=30, color='seagreen', edgecolor='white', alpha=0.7)
plt.axvline(pop_mean, color='red', linewidth=2, label=f'True μ = {pop_mean:.1f}')
plt.axvline(np.mean(sample_means), color='blue', linewidth=2, linestyle='--',
            label=f'Mean of x̄ = {np.mean(sample_means):.1f}')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.title('Distribution of Sample Means\n(Sampling Distribution)')
plt.legend()
plt.show()
```

## Summary

|---------|----------|
